```python
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

tree = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F", "G"],
    "D": [],
    "E": [],
    "F": [],
    "G": []
}

def bfs(tree, root):
    visited = set()
    queue = []

    steps = []

    queue.append(root)

    while queue:
        child = queue.pop(0)

        if child not in visited:
            print(child, end=" ")
            visited.add(child)

            for neighbor in tree[child]:
                if neighbor not in visited:
                    queue.append(neighbor)

            steps.append((child, visited.copy(), queue.copy()))

    return steps


print("BFS Traversal:")
steps = bfs(tree, "A")


G = nx.Graph()

for node in tree:
    for neighbor in tree[node]:
        G.add_edge(node, neighbor)


pos = {
    "A": (0, 3),
    "B": (-1, 2),
    "C": (1, 2),
    "D": (-1.5, 1),
    "E": (-0.5, 1),
    "F": (0.5, 1),
    "G": (1.5, 1)
}


fig, ax = plt.subplots(figsize=(9, 6))


def update(i):
    ax.clear()

    current, visited, queue = steps[i]

    colors = []

    for node in G.nodes():

        if node == current:
            colors.append("orange")

        elif node in visited:
            colors.append("lightgreen")

        elif node in queue:
            colors.append("yellow")

        else:
            colors.append("lightblue")

    nx.draw(
        G,
        pos,
        ax=ax,
        with_labels=True,
        node_color=colors,
        node_size=1800,
        font_size=15,
        font_weight="bold",
        edge_color="black"
    )

    ax.set_title(
        "BFS Visualization - Step " + str(i + 1),
        fontsize=16
    )

    ax.text(
        0, 0.35,
        "Current Node: " + current,
        ha="center",
        fontsize=12
    )

    ax.text(
        0, 0.10,
        "Traversal: " + " -> ".join(
            [step[0] for step in steps[:i+1]]
        ),
        ha="center",
        fontsize=12
    )


ani = FuncAnimation(
    fig,
    update,
    frames=len(steps),
    interval=1200,
    repeat=False
)

plt.close()

display(HTML(ani.to_jshtml()))
```

```python
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

tree = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": ["G", "H"],
    "E": ["I"],
    "F": ["J"],
    "G": [],
    "H": [],
    "I": [],
    "J": []
}

steps = []
visited = []
path = []

def dls(tree, node, goal, depth):

    print(node, end=" ")

    visited.append(node)
    path.append(node)

    steps.append((node, depth, visited.copy(), path.copy()))

    if node == goal:
        path.pop()
        return True

    if depth == 0:
        path.pop()
        return False

    for neighbor in tree[node]:
        if dls(tree, neighbor, goal, depth - 1):
            path.pop()
            return True

    path.pop()
    return False


print("DLS Traversal:")

result = dls(tree, "A", "F", 2)

print("\nGoal Found:", result)


G = nx.DiGraph()

for node in tree:
    for neighbor in tree[node]:
        G.add_edge(node, neighbor)


pos = {
    "A": (0, 3),
    "B": (-1.5, 2),
    "C": (1.5, 2),
    "D": (-2, 1),
    "E": (-1, 1),
    "F": (1.5, 1),
    "G": (-2.5, 0),
    "H": (-1.5, 0),
    "I": (-0.5, 0),
    "J": (1.5, 0)
}


fig, ax = plt.subplots(figsize=(10, 7))


def update(i):

    ax.clear()

    current, depth, visited_nodes, current_path = steps[i]

    colors = []

    for node in G.nodes():

        if node == current:
            colors.append("orange")

        elif node == "F" and node in visited_nodes:
            colors.append("red")

        elif node in current_path:
            colors.append("yellow")

        elif node in visited_nodes:
            colors.append("lightgreen")

        else:
            colors.append("lightblue")

    nx.draw(
        G,
        pos,
        ax=ax,
        with_labels=True,
        node_color=colors,
        node_size=1600,
        font_size=13,
        font_weight="bold",
        edge_color="black",
        arrows=True,
        arrowsize=15
    )

    ax.set_title(
        "Depth Limited Search (DLS) - Step " + str(i + 1),
        fontsize=16
    )

    ax.text(
        0, -0.45,
        "Current Node: " + current +
        "    Remaining Depth: " + str(depth),
        ha="center",
        fontsize=12
    )

    ax.text(
        0, -0.75,
        "Traversal: " + " -> ".join(
            [step[0] for step in steps[:i+1]]
        ),
        ha="center",
        fontsize=11
    )


ani = FuncAnimation(
    fig,
    update,
    frames=len(steps),
    interval=1200,
    repeat=False
)

plt.close()

display(HTML(ani.to_jshtml()))
```

```python
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

tree = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F", "G"],
    "D": ["H"],
    "E": ["I"],
    "F": [],
    "G": ["J"],
    "H": [],
    "I": [],
    "J": []
}

order = []
steps = []
current_limit = 0


def dls(tree, node, goal, depth):

    order.append(node)

    steps.append((
        node,
        current_limit,
        depth,
        order.copy()
    ))

    if node == goal:
        return True

    if depth == 0:
        return False

    for neighbor in tree[node]:

        if dls(tree, neighbor, goal, depth - 1):
            return True

    return False


def ids(tree, start, goal):

    global current_limit

    for depth in range(3):

        order.clear()
        current_limit = depth

        if dls(tree, start, goal, depth):
            return depth


depth = ids(tree, "A", "G")

print("IDS Traversal:")
print(" -> ".join(order))
print("Goal Found: G")
print("Depth Limit:", depth)


G = nx.DiGraph()

for node in tree:
    for neighbor in tree[node]:
        G.add_edge(node, neighbor)


pos = {
    "A": (0, 4),
    "B": (-1.5, 3),
    "C": (1.5, 3),
    "D": (-2, 2),
    "E": (-1, 2),
    "F": (1, 2),
    "G": (2, 2),
    "H": (-2, 1),
    "I": (-1, 1),
    "J": (2, 1)
}


fig, ax = plt.subplots(figsize=(10, 7))


def update(i):

    ax.clear()

    current, limit, remaining, visited = steps[i]

    colors = []

    for node in G.nodes():

        if node == current and node == "G":
            colors.append("red")

        elif node == current:
            colors.append("orange")

        elif node in visited:
            colors.append("lightgreen")

        else:
            colors.append("lightblue")

    nx.draw(
        G,
        pos,
        ax=ax,
        with_labels=True,
        node_color=colors,
        node_size=1500,
        font_size=13,
        font_weight="bold",
        edge_color="black",
        arrows=True,
        arrowsize=15
    )

    ax.set_title(
        "IDS Visualization - Depth Limit: " + str(limit),
        fontsize=16
    )

    ax.text(
        0, 0.45,
        "Current Node: " + current +
        "    Remaining Depth: " + str(remaining),
        ha="center",
        fontsize=12
    )

    ax.text(
        0, 0.15,
        "Traversal: " + " -> ".join(visited),
        ha="center",
        fontsize=11
    )


ani = FuncAnimation(
    fig,
    update,
    frames=len(steps),
    interval=1200,
    repeat=False
)

plt.close()

display(HTML(ani.to_jshtml()))
```

```python
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

graph = {
    "A": [("B", 9), ("C", 4)],
    "B": [("C", 2), ("D", 7), ("E", 3)],
    "C": [("D", 1), ("E", 6)],
    "D": [("E", 4), ("F", 8)],
    "E": [("F", 2)],
    "F": [("D", 8), ("E", 2)]
}

start = "A"
goal = "F"

queue = [(0, start, [start])]

steps = []
result_path = []
total_cost = 0


while queue:

    queue.sort()

    cost, node, path = queue.pop(0)

    steps.append((
        cost,
        node,
        path.copy(),
        queue.copy()
    ))

    if node == goal:
        print("Path:", " -> ".join(path))
        print("Total Cost =", cost)

        result_path = path.copy()
        total_cost = cost
        break

    for neighbor, edge_cost in graph[node]:

        new_cost = cost + edge_cost
        new_path = path + [neighbor]

        queue.append((new_cost, neighbor, new_path))


G = nx.DiGraph()

for node in graph:
    for neighbor, edge_cost in graph[node]:
        G.add_edge(node, neighbor, weight=edge_cost)



pos = {
    "A": (0, 2),
    "B": (-2, 1),
    "C": (0, 1),
    "D": (2, 1),
    "E": (-1, 0),
    "F": (1, 0)
}



fig, ax = plt.subplots(figsize=(10, 7))


def update(i):

    ax.clear()

    cost, current, path, queue_now = steps[i]

    colors = []

    for node in G.nodes():

        if node == current:
            colors.append("orange")

        elif node in path:
            colors.append("lightgreen")

        else:
            colors.append("lightblue")


    nx.draw(
        G,
        pos,
        ax=ax,
        with_labels=True,
        node_color=colors,
        node_size=1800,
        font_size=14,
        font_weight="bold",
        edge_color="gray",
        arrows=True,
        arrowsize=18
    )


    edge_labels = nx.get_edge_attributes(G, "weight")

    nx.draw_networkx_edge_labels(
        G,
        pos,
        edge_labels=edge_labels,
        ax=ax,
        font_size=10
    )


    if result_path:

        path_edges = list(zip(result_path, result_path[1:]))

        nx.draw_networkx_edges(
            G,
            pos,
            edgelist=path_edges,
            ax=ax,
            edge_color="green",
            width=3,
            arrows=True,
            arrowsize=20
        )

    ax.set_title(
        "Uniform Cost Search (UCS) - Step " + str(i + 1),
        fontsize=16
    )

    ax.text(
        0, -0.5,
        "Current Node: " + current +
        "    Current Cost: " + str(cost),
        ha="center",
        fontsize=12
    )

    ax.text(
        0, -0.8,
        "Current Path: " + " -> ".join(path),
        ha="center",
        fontsize=11
    )



ani = FuncAnimation(
    fig,
    update,
    frames=len(steps),
    interval=1500,
    repeat=False
)

plt.close()

display(HTML(ani.to_jshtml()))
```
